Now, you can take a look at what are the coordinates for each site.


```python
# Select relevant sub-columns: `Launch Site`, `Lat(Latitude)`, `Long(Longitude)`, `class`
spacex_df = spacex_df[['Launch Site', 'Lat', 'Long', 'class']]
launch_sites_df = spacex_df.groupby(['Launch Site'], as_index=False).first()
launch_sites_df = launch_sites_df[['Launch Site', 'Lat', 'Long']]
launch_sites_df
```

**Findings (Task 1):**

* **Are all launch sites in proximity to the Equator line?** No, none of them are actually near the Equator. The Florida sites (CCAFS LC-40, CCAFS SLC-40, KSC LC-39A) sit at roughly 28.5°N and VAFB SLC-4E in California at roughly 34.6°N. However, they are all placed as far south as is practical within the continental US. Launching closer to the Equator gives a rocket a larger "free" eastward velocity from Earth's rotation (about 1,670 km/h at the Equator), which saves fuel, particularly for missions to equatorial and geostationary orbits. This is one reason Florida is the main US launch hub.
* **Are all launch sites in very close proximity to the coast?** Yes. All four sites are within a few kilometres of the coastline: the Florida sites face the Atlantic Ocean to the east and VAFB faces the Pacific Ocean to the west. Launching over open water means that spent stages and debris from a failed launch fall into the sea rather than onto populated areas. East-facing coasts (Florida) suit eastward launches that benefit from Earth's rotation, while VAFB's west/south-facing coast suits polar and sun-synchronous orbits.

Next, let's try to enhance the map by adding the launch outcomes for each site, and see which sites have high success rates.

Next, let's create markers for all launch records.
If a launch was successful `(class=1)`, then we use a green marker and if a launch was failed, we use a red marker `(class=0)`


Note that a launch only happens in one of the four launch sites, which means many launch records will have the exact same coordinate. Marker clusters can be a good way to simplify a map containing many markers having the same coordinate.


Let's first create a `MarkerCluster` object


```python
marker_cluster = MarkerCluster()
```

*TODO:* Create a new column in `spacex_df` dataframe called `marker_color` to store the marker colors based on the `class` value


```python

# Apply a function to check the value of `class` column
# If class=1, marker_color value will be green
# If class=0, marker_color value will be red
def assign_marker_color(launch_class):
    return 'green' if launch_class == 1 else 'red'

spacex_df['marker_color'] = spacex_df['class'].apply(assign_marker_color)
spacex_df.tail(10)
```

*TODO:* For each launch result in `spacex_df` data frame, add a `folium.Marker` to `marker_cluster`


```python
# Add marker_cluster to current site_map
site_map.add_child(marker_cluster)

# for each row in spacex_df data frame
# create a Marker object with its coordinate
# and customize the Marker's icon property to indicate if this launch was successed or failed, 
# e.g., icon=folium.Icon(color='white', icon_color=row['marker_color']
for index, record in spacex_df.iterrows():
    # Create and add a Marker to the marker cluster
    marker = folium.Marker(
        location=[record['Lat'], record['Long']],
        icon=folium.Icon(color='white', icon_color=record['marker_color']),
        popup=f"{record['Launch Site']} - {'Success' if record['class'] == 1 else 'Failure'}",
    )
    marker_cluster.add_child(marker)

site_map
```

Next, we need to explore and analyze the proximities of launch sites.


Let's first add a `MousePosition` on the map to get coordinate for a mouse over a point on the map. As such, while you are exploring the map, you can easily find the coordinates of any points of interests (such as railway)


```python
# Add Mouse Position to get the coordinate (Lat, Long) for a mouse over on the map
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)
site_map
```

Now zoom in to a launch site and explore its proximity to see if you can easily find any railway, highway, coastline, etc. Move your mouse to these points and mark down their coordinates (shown on the top-left) in order to the distance to the launch site.

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

*TODO:* Mark down a point on the closest coastline using MousePosition and calculate the distance between the coastline point and the launch site.


```python
# find coordinate of the closet coastline
# e.g.,: Lat: 28.56367  Lon: -80.57163
# distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)

# Launch site used for the proximity analysis: CCAFS SLC-40
launch_site_name = 'CCAFS SLC-40'
launch_site_lat = launch_sites_df.loc[launch_sites_df['Launch Site'] == launch_site_name, 'Lat'].values[0]
launch_site_lon = launch_sites_df.loc[launch_sites_df['Launch Site'] == launch_site_name, 'Long'].values[0]

# Closest coastline point (read from MousePosition)
coastline_lat = 28.56367
coastline_lon = -80.57163

distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
print(f"Distance from {launch_site_name} to the coastline: {distance_coastline:.2f} km")
```

```python
# Create and add a folium.Marker on your selected closest coastline point on the map
# Display the distance between coastline point and launch site using the icon property 
coastline_coordinate = [coastline_lat, coastline_lon]
distance_marker = folium.Marker(
    coastline_coordinate,
    icon=DivIcon(
        icon_size=(20, 20),
        icon_anchor=(0, 0),
        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance_coastline),
    )
)
site_map.add_child(distance_marker)
```

*TODO:* Draw a `PolyLine` between a launch site to the selected coastline point


```python
# Create a `folium.PolyLine` object using the coastline coordinates and launch site coordinate
coordinates = [[launch_site_lat, launch_site_lon], coastline_coordinate]
lines = folium.PolyLine(locations=coordinates, weight=1)
site_map.add_child(lines)
site_map
```

```python
# Create a marker with distance to a closest city, railway, highway, etc.
# Draw a line between the marker to the launch site

def add_proximity(site_map, site_coord, point_coord, label, color='#d35400'):
    """Add a distance label at point_coord and a line back to the launch site."""
    distance = calculate_distance(site_coord[0], site_coord[1], point_coord[0], point_coord[1])
    folium.Marker(
        point_coord,
        icon=DivIcon(
            icon_size=(20, 20),
            icon_anchor=(0, 0),
            html='<div style="font-size: 12; color:%s;"><b>%s</b></div>' % (color, "{:10.2f} KM".format(distance)),
        ),
        tooltip=label,
    ).add_to(site_map)
    folium.PolyLine(locations=[site_coord, point_coord], weight=1, color=color, tooltip=label).add_to(site_map)
    print(f"{label}: {distance:.2f} km")
    return distance

launch_site_coordinate = [launch_site_lat, launch_site_lon]

# Coordinates read from MousePosition (approximate; adjust if you pick different points)
railway_coordinate = [28.57208, -80.58527]   # NASA railway west of the site
highway_coordinate = [28.56335, -80.57085]   # Samuel C. Phillips Parkway
city_coordinate    = [28.40150, -80.60440]   # Cape Canaveral (city)
```

```python
distance_railway = add_proximity(site_map, launch_site_coordinate, railway_coordinate, 'Closest railway', color='#2980b9')
distance_highway = add_proximity(site_map, launch_site_coordinate, highway_coordinate, 'Closest highway', color='#8e44ad')
distance_city    = add_proximity(site_map, launch_site_coordinate, city_coordinate, 'Closest city (Cape Canaveral)', color='#c0392b')
```

```python
# Summary of distances from CCAFS SLC-40 and the final map
proximity_df = pd.DataFrame({
    'Proximity': ['Coastline', 'Railway', 'Highway', 'City (Cape Canaveral)'],
    'Distance (km)': [distance_coastline, distance_railway, distance_highway, distance_city],
}).round(2)
display(proximity_df)

site_map
```

**Findings (Task 3), using CCAFS SLC-40:**

* **Are launch sites in close proximity to railways?** Yes. The NASA railway is only about 1.3 km away. Rail access is needed to move heavy rocket stages, equipment and propellant to the site.
* **Are launch sites in close proximity to highways?** Yes. The nearest highway (Samuel C. Phillips Parkway) is under 1 km away, giving road access for staff, equipment and emergency services.
* **Are launch sites in close proximity to coastline?** Yes. The coastline is under 1 km away, so rockets fly over the ocean immediately after lift-off and any debris or failed stages land in the sea.
* **Do launch sites keep certain distance away from cities?** Yes. The nearest city (Cape Canaveral) is roughly 18 km away. Launch sites are kept well away from populated areas to limit the danger from explosions, noise and falling debris, while still being close enough for the workforce to commute.

In short, launch sites are placed close to transport infrastructure (railways, highways) and the coast for logistics and safety, but far from cities to protect the public. They are also located as far south as practical to gain extra velocity from Earth's rotation.
